# Python Assignment1: Data Structure - Strings and Tuples 

## 1. String Concatenation

```python
string1="Hello"
string2=input("Enter Your name:")
string=string1+' '+string2
print(string)
string3=", welcome to Python Programming"
string=string+string3
print(string)
```

## 2.String Slicing and Indexing

```python
a=string
print(a)
print(a[0])
print(a[-1])
print(a[:5])
print(a[-11:])
print(a[::-1])
print(a[-18:-12:])
```

## 3. String Methods

```python
strM="Python beginner tutorial"
print(strM.upper())
print(strM.lower())
print(strM.capitalize())
print(strM.count("t"))
print(strM.replace("Python","Machine Learning"))
```

### Tuples

```python
t1=(10,20,30)
t2=(40,50,60)
t_combine=t1+t2
print(t_combine)
print(t_combine*3)
print(t_combine[2])
print(t_combine[:3])
print(t_combine[-3:])
```
