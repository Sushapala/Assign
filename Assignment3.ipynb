### Number Guessing Game

```python
import random
i=1
secret_num=random.randint(1,10)
while i<=3:
    n=int(input("Guess the number between 1 and 10:"))
    if n>10 or n<=0:
        print("Your guess is out of range. Please guess a number between 1 and 10")
        continue
    elif n>secret_num:
        print("Your guess is too high. Try again")
    elif n<secret_num:
        print("Your guess is too low. Try again")
    elif n==secret_num:
        print("Congratulations! You guess the correct number")
        break
    i=i+1
else:
    print("Better luck next time!")  
```

### Multiplication Table Generator

```python
num=int(input("Enter a number for which you want the multiplication table"))
for i in range(1,11):
    print(num,"*",i,"=",num*i)
```

### BMI Calculator (Using function)

```python
def calculate_bmi(weight,height):
    bmi=round(weight/height**2,2)
    return bmi

w=float(input("Enter your weight in kg:"))
h=float(input("Enter your height in meters:"))
print("Your BMI is - ",calculate_bmi(w,h))
```
